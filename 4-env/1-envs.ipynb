# Managing the environment

## Project initialization

Create a project: a dedicated space where we can manage functions, artifacts and executions.

Use the ``username`` as param to create a personal project in the shared space.

```python
import digitalhub as dh
import os

project = dh.get_or_create_project(f"{os.environ['USER']}-base")
project
```

## 1. Create a dummy function 

We define a simple function for printing the environment to stdout, to verify everything is set up correctly.



```python
func = project.new_function(name="printenv",
                            kind="container",
                            command="printenv",
                            image="alpine:latest"
                            )
```

Now we should be able to run it

```python
run = func.run("job", local_execution=False, run_as_user="8877", wait=True)
```

Check the logs.

```python
print(run.logs()[0]['content'])
```

### 1.2 Setting custom env variables

Add env to the run execution and verify they are injected in the env.

```python
run = func.run("job", local_execution=False, run_as_user="8877", envs=[{"name": "MY_ENV_VAR", "value": "plaintext value"}], wait=True)
```

```python
print(run.logs()[0]['content'])
```

Env are also visible in the spec, check the export

```python
print(run.spec.envs)
```

### 1.3 Secrets

Secrets can be used to keep the value sealed until execution time.
Create a secret named 'TEST_KEY' with a value via console, and then we can use for the run.




```python
run = func.run("job", local_execution=False, run_as_user="8877", secrets=["TEST_KEY"], wait=True)
```

Check if the log contains the value.
